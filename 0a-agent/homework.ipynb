```python
# Generating Random Numbers
import random

known_weather_data = {
    'berlin': 20.0
}

def get_weather(city: str) -> float:
    city = city.strip().lower()

    if city in known_weather_data:
        return known_weather_data[city]

    return round(random.uniform(-5, 35), 1)
```

```python
# Function Tools
get_weather_tool = {
    "type": "function",
    "name": "get_weather",
    "description": "Returns the current temperature for a given city",
    "parameters": {
        "type": "object",
        "properties": {
            "city": {
                "type": "string",
                "description": "The name of the city to get weather data for."
            }
        },
        "required": ["city"],
        "additionalProperties": False
    }
}
```

```python
def set_weather(city: str, temp: float) -> None:
    city = city.strip().lower()
    known_weather_data[city] = temp
    return 'OK'
```

```python
set_weather_tool = {
    "type": "function",
    "name": "set_weather",
    "description": "Sets or updates the temperature for a specific city in the weather database.",
    "parameters": {
        "type": "object",
        "properties": {
            "city": {
                "type": "string",
                "description": "The name of the city to set the weather for."
            },
            "temp": {
                "type": "number",
                "description": "The temperature to associate with the city in degrees Celsius."
            }
        },
        "required": ["city", "temp"],
        "additionalProperties": False
    }
}
```

```python
# Simple MCP Server
# weather_server.py
from fastmcp import FastMCP

mcp = FastMCP("Demo 🚀")

@mcp.tool
def add(a: int, b: int) -> int:
    """Add two numbers"""
    return a + b

# Use async-friendly version in Colab
await mcp.run_async()
```
