```python
# import required libs
from minsearch import VectorSearch
import numpy as np
import pandas as pd
from sentence_transformers import SentenceTransformer
```

```python
# 1. Prepare your documents
documents = [
    {
        "id": 1,
        "title": "Introduction to Data Science",
        "description": "Learn the basics of data science, including data cleaning, analysis, and visualization. This course is for beginners.",
        "category": "Education"
    },
    {
        "id": 2,
        "title": "Machine Learning for Beginners",
        "description": "Explore fundamental machine learning algorithms like linear regression, decision trees, and clustering. Hands-on exercises included.",
        "category": "Education"
    },
    {
        "id": 3,
        "title": "Python Programming Masterclass",
        "description": "Master Python from scratch. Covers core concepts, object-oriented programming, and popular libraries like Pandas and NumPy.",
        "category": "Programming"
    },
    {
        "id": 4,
        "title": "Advanced SQL Techniques",
        "description": "Deep dive into SQL for data analysis. Learn complex queries, window functions, and database optimization.",
        "category": "Data Engineering"
    },
    {
        "id": 5,
        "title": "Understanding Artificial Intelligence",
        "description": "A general overview of AI, its history, applications, and ethical considerations. No programming required.",
        "category": "Education"
    },
    {
        "id": 6,
        "title": "Building Web Applications with Flask",
        "description": "Learn to create dynamic web applications using the Flask framework in Python. Includes database integration.",
        "category": "Programming"
    }
]
```

```python
# Loading Embedding Model
print("Loading Embedding Transformer Model")
model = SentenceTransformer('all-MiniLM-L6-v2')
print("Model Successfully Loaded")
```

```python
document_texts = [f"{doc['title']} {doc['description']}" for doc in documents]
doc_embed = model.encode(document_texts)
print("Embedding Successful!")
```

```python
print(f"Generated {len(doc_embed)} embeddings, each with {doc_embed.shape[1]} dimensions.") # -> Cols
```

```python
payload = documents
vector_index = VectorSearch(keyword_fields=["category"])
```

```python
print("Fitting the documents")
vector_index.fit(doc_embed, payload)
print("Vector Index Fitted!")
```

```python
query_1 = "Courses about programming languages"
query_embed = model.encode([query_1])[0]
```

```python
print(f"Searching for {query_1}")
results_1 = vector_index.search(query_embed, num_results=3)
```

```python
scores = []
print("Search Results:")
for res in results_1:
    print(res)
    scores.append(res['category'])
    
```
